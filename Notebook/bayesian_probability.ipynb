# Bayes' Theorem

Bayes' Theorem is a mathematical formula that updates the probability of a hypothesis after observing new evidence.

For this project, we compute:

- Prior:
  
  P(Positive)

- Likelihood:

  P(keyword | Positive)

- Marginal:

  P(keyword)

- Posterior:

  P(Positive | keyword)

Bayes' Theorem:

P(Positive | keyword) =
(P(keyword | Positive) × P(Positive))
/
P(keyword)

Interpretation:

The posterior probability tells us how likely a review is to be positive after observing a specific keyword.
